```python
players_and_coaches_df = pd.read_csv('../downloaded_files/scrape_exports/afc_match_lineups_coaches.csv')
print("Players Head:")
players_and_coaches_df.head()
```

```python
# 1. Filter results to AFC matches since 1995 (Ensure afc_matches_since_1995 is ready)
# (Actually, the user mentioned afc_matches_since_2000 in the prompt but then said 1995 in the description. 
# I will use 1995 as the filter for the final DF as requested in the text.)

afc_matches_since_1995 = df[
    (df['home_team'].isin(teams_in_asia)) & 
    (df['away_team'].isin(teams_in_asia)) & 
    (df['date'].dt.year >= 1995)
].copy()

# Add match_result to afc_matches_since_1995
def get_match_result(row, team_type):
    if row['home_score'] == row['away_score']:
        return 'Draw'
    if team_type == 'home':
        return 'Win' if row['home_score'] > row['away_score'] else 'Loss'
    else:
        return 'Win' if row['away_score'] > row['home_score'] else 'Loss'

# 2. Prepare players_and_coaches_df
# Convert date to datetime to match
players_and_coaches_df['date'] = pd.to_datetime(players_and_coaches_df['date'])

# 3. Merge with results to get play_against, location, and match_result
# We need to merge twice: once for home and once for away

# Home players
home_merge = pd.merge(
    players_and_coaches_df,
    afc_matches_since_1995,
    left_on=['date', 'team'],
    right_on=['date', 'home_team']
)
home_merge['play_against'] = home_merge['away_team']
home_merge['location'] = home_merge['neutral'].apply(lambda x: 'neutral' if x else 'home')
home_merge['match_result'] = home_merge.apply(lambda r: get_match_result(r, 'home'), axis=1)

# Away players
away_merge = pd.merge(
    players_and_coaches_df,
    afc_matches_since_1995,
    left_on=['date', 'team'],
    right_on=['date', 'away_team']
)
away_merge['play_against'] = away_merge['home_team']
away_merge['location'] = away_merge['neutral'].apply(lambda x: 'neutral' if x else 'away')
away_merge['match_result'] = away_merge.apply(lambda r: get_match_result(r, 'away'), axis=1)

# Combine
players_and_coaches_df_full = pd.concat([home_merge, away_merge], ignore_index=True)

# Select and rename columns as requested
cols_to_keep = [
    'team', 'player_name', 'coach', 'play_against', 'location', 
    'date', 'match_result', 'goals', 'assists', 'yellow_cards', 'red_cards', 'match_id'
]
players_and_coaches_df_full = players_and_coaches_df_full[cols_to_keep].rename(columns={'date': 'date_played'})

# 4. Identify WIP matches
# Matches in AFC but not represented in the lineup table
# All matches where (home AND away in teams_in_asia) AND (year >= 1995)
# BUT we check which of these don't have ANY players in players_and_coaches_df_full

scraped_match_ids = set(players_and_coaches_df['match_id'].unique())
# However, afc_matches_since_1995 doesn't have match_id. 
# We need to define match_id in afc_matches_since_1995 to compare.
# Or use (date, home_team, away_team) as key.

# Let's create a temporary key for comparison
afc_matches_since_1995['temp_key'] = afc_matches_since_1995['date'].astype(str) + "_" + afc_matches_since_1995['home_team'] + "_" + afc_matches_since_1995['away_team']
players_and_coaches_df_full['temp_key'] = players_and_coaches_df_full['date_played'].astype(str) + "_" + \
    players_and_coaches_df_full.apply(lambda x: x['team'] if x['location'] in ['home', 'neutral'] else x['play_against'], axis=1) + "_" + \
    players_and_coaches_df_full.apply(lambda x: x['play_against'] if x['location'] in ['home', 'neutral'] else x['team'], axis=1)

# Actually, the user asked for players_and_coaches_df_WIP for games not represented IN THIS TABLE.
# This implies games from afc_matches_since_1995 that are NOT in players_and_coaches_df_full.

matched_keys = set(players_and_coaches_df_full['temp_key'].unique())
players_and_coaches_df_WIP = afc_matches_since_1995[~afc_matches_since_1995['temp_key'].isin(matched_keys)].copy()

# Cleanup
afc_matches_since_1995.drop(columns=['temp_key'], inplace=True)
players_and_coaches_df_full.drop(columns=['temp_key'], inplace=True)

print(f"Full dataframe shape: {players_and_coaches_df_full.shape}")
print(f"WIP dataframe shape: {players_and_coaches_df_WIP.shape}")
players_and_coaches_df_full
```
